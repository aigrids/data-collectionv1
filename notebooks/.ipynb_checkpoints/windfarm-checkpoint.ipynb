# Benchmarks, analysis and inspection of the WindFarm task and datasets

Given two weeks of historical power generation measurements for each turbine in a wind farm, together with the relative geographic positions of the turbines, the goal of the WindFarm task is to predict the active power output profile of the wind farm over the subsequent 48-hour forecasting horizon.

Accurate forecasting of wind farm power generation enables power system operators to better manage the inherent variability and uncertainty of wind energy production. Improved forecasts support the efficient utilization of energy storage systems and flexible demand resources, enhance the planning and dispatch of transmission networks, and ultimately facilitate the reliable integration of larger shares of wind energy into existing power grids.

The WindFarm task is based on multivariate spatio-temporal time-series data, consisting of historical turbine-level power generation measurements together with the relative geographic positions of turbines within the wind farm.

The WindFarm task is formulated as a supervised learning problem, where models are trained to predict future wind power generation profiles from historical turbine-level measurements and spatial information describing turbine locations within the wind farm. The task corresponds to a multi-step spatio-temporal forecasting problem, in which models must learn both temporal dependencies in wind power generation and spatial interactions among turbines induced by wind flow dynamics.

We define six forecasting sub-task datasets grouped according to two prediction horizons: forecasting wind power generation 48 hours ahead and forecasting wind power generation 72 hours ahead. Within each prediction horizon, the sub-tasks are designed to evaluate temporal, spatial, and spatio-temporal generalization capabilities of forecasting models.
