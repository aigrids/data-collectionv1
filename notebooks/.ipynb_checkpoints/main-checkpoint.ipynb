```python
import os
import sys

sys.path.append('../')
sys.path.append('../src')
sys.path.append('../ai4climate')
import configuration
import taskdata
from mlmodel import PowerFoundationModel
import preprocess
import training
import plot

PATH_CONFIG = '../config.yml'
PATH_RESULTS = '../results/'
PATH_DATA_ROOT = '/scratch/AI4Climate/'
```

```python
# parse experiment configurations into dictionary
cfg = configuration.ExperimentConfiguration(PATH_CONFIG, PATH_RESULTS)

# load task datasets
taskdata_dict = taskdata.load_all(cfg, PATH_DATA_ROOT)

# preprocess datasets
taskdata_dict = preprocess.build_dataloaders(cfg, taskdata_dict)

# get a sample data point for model design
sample_datapoint_batch = next(
    iter(taskdata_dict['buildingelectricity']['train_data'])
)

# build foundation model
fmodel = PowerFoundationModel(cfg, sample_datapoint_batch)

# train model
results = training.train_model(cfg, taskdata_dict, fmodel)

# plot resulting loss curves
plot.plot_loss_curves(cfg, results)
```
